```python
def subtract(a, b):
    output = a - b
    return output
```

```python
subtract(10, 5)
```

```python
subtract(10,3)
```

```python
subtract(5,10)
```

```python
def distance_from_zero(x):
    if x < 0:
        return subtract(0, x)
    else:
        return subtract(x, 0)
```

```python
distance_from_zero(-5)
```
