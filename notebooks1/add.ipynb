```python
def sumlist(lst):
    total = 0
    for num in lst:
        total += num
    return total
```

```python
sumlist([1,2,3,4,5])
```

```python
sumlist([3,2,1,4,9])
```

```python
sumlist([1,2,3,44,5])
```
